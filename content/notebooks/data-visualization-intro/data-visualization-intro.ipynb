# Introduction to Data Visualization

Learn to create beautiful charts and graphs with matplotlib! This notebook covers the fundamentals of data visualization in Python.

**What you'll learn:**
- Line plots
- Bar charts
- Scatter plots
- Customizing plots
- Multiple subplots

```python
# First, import the libraries we need
import matplotlib.pyplot as plt
import numpy as np

# Enable inline plotting
%matplotlib inline

print("Libraries loaded! ✓")
```

## 1. Line Plots

Line plots are great for showing trends over time.

```python
# Simple line plot
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.figure(figsize=(8, 5))
plt.plot(x, y)
plt.title('My First Line Plot')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Multiple lines with styling
x = np.linspace(0, 10, 100)

plt.figure(figsize=(10, 6))
plt.plot(x, np.sin(x), label='sin(x)', color='blue', linewidth=2)
plt.plot(x, np.cos(x), label='cos(x)', color='red', linestyle='--', linewidth=2)
plt.plot(x, np.sin(x) + np.cos(x), label='sin(x) + cos(x)', color='green', linestyle=':', linewidth=2)

plt.title('Trigonometric Functions', fontsize=14, fontweight='bold')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 2. Bar Charts

Bar charts are perfect for comparing categories.

```python
# Simple bar chart
categories = ['Python', 'JavaScript', 'Java', 'C++', 'Go']
popularity = [85, 78, 65, 55, 45]
colors = ['#3776ab', '#f7df1e', '#ed8b00', '#00599C', '#00ADD8']

plt.figure(figsize=(10, 6))
plt.bar(categories, popularity, color=colors, edgecolor='black')
plt.title('Programming Language Popularity', fontsize=14, fontweight='bold')
plt.xlabel('Language')
plt.ylabel('Popularity Score')
plt.ylim(0, 100)

# Add value labels on bars
for i, v in enumerate(popularity):
    plt.text(i, v + 2, str(v), ha='center', fontweight='bold')

plt.show()
```

```python
# Grouped bar chart
labels = ['Q1', 'Q2', 'Q3', 'Q4']
product_a = [20, 35, 30, 45]
product_b = [25, 32, 34, 40]

x = np.arange(len(labels))
width = 0.35

plt.figure(figsize=(10, 6))
bars1 = plt.bar(x - width/2, product_a, width, label='Product A', color='steelblue')
bars2 = plt.bar(x + width/2, product_b, width, label='Product B', color='coral')

plt.title('Quarterly Sales Comparison', fontsize=14, fontweight='bold')
plt.xlabel('Quarter')
plt.ylabel('Sales (thousands)')
plt.xticks(x, labels)
plt.legend()
plt.show()
```

## 3. Scatter Plots

Scatter plots show relationships between two variables.

```python
# Random scatter plot
np.random.seed(42)
x = np.random.randn(100)
y = 2 * x + np.random.randn(100) * 0.5

plt.figure(figsize=(8, 6))
plt.scatter(x, y, alpha=0.7, c='purple', edgecolors='black', s=60)
plt.title('Scatter Plot with Correlation', fontsize=14, fontweight='bold')
plt.xlabel('X Variable')
plt.ylabel('Y Variable')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Scatter plot with color mapping
np.random.seed(42)
n = 100
x = np.random.rand(n)
y = np.random.rand(n)
sizes = np.random.rand(n) * 500  # Point sizes
colors = np.random.rand(n)        # Color values

plt.figure(figsize=(10, 7))
scatter = plt.scatter(x, y, c=colors, s=sizes, alpha=0.6, cmap='viridis', edgecolors='black')
plt.colorbar(scatter, label='Color Value')
plt.title('Bubble Chart', fontsize=14, fontweight='bold')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

## 4. Pie Charts

Pie charts show parts of a whole.

```python
# Pie chart
labels = ['Desktop', 'Mobile', 'Tablet', 'Other']
sizes = [45, 40, 10, 5]
explode = (0, 0.1, 0, 0)  # Explode the Mobile slice
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']

plt.figure(figsize=(8, 8))
plt.pie(sizes, explode=explode, labels=labels, colors=colors,
        autopct='%1.1f%%', shadow=True, startangle=90)
plt.title('Website Traffic by Device', fontsize=14, fontweight='bold')
plt.axis('equal')  # Equal aspect ratio for circular pie
plt.show()
```

## 5. Histograms

Histograms show the distribution of data.

```python
# Histogram
np.random.seed(42)
data = np.random.normal(100, 15, 1000)  # Mean=100, Std=15, 1000 samples

plt.figure(figsize=(10, 6))
plt.hist(data, bins=30, color='steelblue', edgecolor='black', alpha=0.7)
plt.axvline(data.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {data.mean():.1f}')
plt.title('Distribution of Test Scores', fontsize=14, fontweight='bold')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

## 6. Subplots

Create multiple plots in one figure.

```python
# 2x2 grid of subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Multiple Visualizations', fontsize=16, fontweight='bold')

# Plot 1: Line
x = np.linspace(0, 10, 100)
axes[0, 0].plot(x, np.sin(x), 'b-', linewidth=2)
axes[0, 0].set_title('Sine Wave')
axes[0, 0].grid(True, alpha=0.3)

# Plot 2: Bar
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 56, 78]
axes[0, 1].bar(categories, values, color=['red', 'green', 'blue', 'orange'])
axes[0, 1].set_title('Bar Chart')

# Plot 3: Scatter
np.random.seed(42)
axes[1, 0].scatter(np.random.rand(50), np.random.rand(50), c='purple', alpha=0.6, s=100)
axes[1, 0].set_title('Scatter Plot')

# Plot 4: Histogram
axes[1, 1].hist(np.random.randn(1000), bins=25, color='teal', edgecolor='black')
axes[1, 1].set_title('Histogram')

plt.tight_layout()
plt.show()
```

## 🎯 Practice: Create Your Own Visualization

Try creating a visualization with this sample data!

```python
# Sample data: Monthly temperatures
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
temperatures = [32, 35, 45, 55, 65, 75, 82, 80, 72, 58, 45, 35]

# Create your visualization here!
plt.figure(figsize=(12, 6))
plt.plot(months, temperatures, 'o-', color='coral', linewidth=2, markersize=8)
plt.fill_between(months, temperatures, alpha=0.3, color='coral')
plt.title('Monthly Average Temperatures', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Temperature (°F)')
plt.grid(True, alpha=0.3)
plt.show()
```
