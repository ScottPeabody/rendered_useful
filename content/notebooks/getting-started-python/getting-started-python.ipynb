## 1. Variables and Data Types

Variables store data. Python automatically detects the type.

```python
# Strings - text data
name = "Alice"
greeting = 'Hello, World!'

# Numbers
age = 25           # Integer
height = 5.7       # Float (decimal)

# Boolean - True or False
is_student = True

# Print them all
print(f"Name: {name}")
print(f"Age: {age}")
print(f"Height: {height}")
print(f"Is student: {is_student}")
```

```python
# Check the type of a variable
print(type(name))       # <class 'str'>
print(type(age))        # <class 'int'>
print(type(height))     # <class 'float'>
print(type(is_student)) # <class 'bool'>
```

## 2. Basic Operations

Python supports all standard math operations.

```python
a = 10
b = 3

print(f"Addition: {a} + {b} = {a + b}")
print(f"Subtraction: {a} - {b} = {a - b}")
print(f"Multiplication: {a} * {b} = {a * b}")
print(f"Division: {a} / {b} = {a / b}")
print(f"Integer Division: {a} // {b} = {a // b}")
print(f"Modulo (remainder): {a} % {b} = {a % b}")
print(f"Power: {a} ** {b} = {a ** b}")
```

```python
# String operations
first_name = "John"
last_name = "Doe"

# Concatenation
full_name = first_name + " " + last_name
print(f"Full name: {full_name}")

# Repetition
print("Ha" * 3)  # HaHaHa

# Length
print(f"Name length: {len(full_name)}")
```

## 3. Control Flow

### If/Else Statements
Make decisions in your code.

```python
temperature = 22

if temperature > 30:
    print("It's hot! 🔥")
elif temperature > 20:
    print("It's nice weather! ☀️")
elif temperature > 10:
    print("It's cool 🍂")
else:
    print("It's cold! ❄️")
```

```python
# Comparison operators
x = 5
y = 10

print(f"{x} == {y}: {x == y}")  # Equal
print(f"{x} != {y}: {x != y}")  # Not equal
print(f"{x} < {y}: {x < y}")    # Less than
print(f"{x} > {y}: {x > y}")    # Greater than
print(f"{x} <= {y}: {x <= y}")  # Less than or equal
print(f"{x} >= {y}: {x >= y}")  # Greater than or equal
```

### Loops
Repeat actions multiple times.

```python
# For loop - iterate over a sequence
fruits = ["apple", "banana", "cherry"]

print("My favorite fruits:")
for fruit in fruits:
    print(f"  - {fruit}")
```

```python
# Range - generate numbers
print("Counting to 5:")
for i in range(1, 6):
    print(i)
```

```python
# While loop - repeat while condition is true
countdown = 5

print("Countdown:")
while countdown > 0:
    print(countdown)
    countdown -= 1
print("Liftoff! 🚀")
```

## 4. Functions

Functions are reusable blocks of code.

```python
# Define a function
def greet(name):
    """Return a greeting message."""
    return f"Hello, {name}! 👋"

# Call the function
message = greet("Alice")
print(message)
print(greet("Bob"))
```

```python
# Function with default parameter
def power(base, exponent=2):
    """Calculate base raised to exponent."""
    return base ** exponent

print(f"3^2 = {power(3)}")
print(f"2^8 = {power(2, 8)}")
```

```python
# Function returning multiple values
def get_stats(numbers):
    """Return min, max, and average of a list."""
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

data = [4, 8, 15, 16, 23, 42]
minimum, maximum, average = get_stats(data)

print(f"Data: {data}")
print(f"Min: {minimum}, Max: {maximum}, Avg: {average:.2f}")
```

## 5. Lists

Lists store ordered collections of items.

```python
# Create a list
colors = ["red", "green", "blue"]
print(f"Colors: {colors}")

# Access by index (0-based)
print(f"First color: {colors[0]}")
print(f"Last color: {colors[-1]}")

# Modify
colors[1] = "yellow"
print(f"Modified: {colors}")
```

```python
# List methods
numbers = [3, 1, 4, 1, 5]
print(f"Original: {numbers}")

numbers.append(9)      # Add to end
print(f"After append: {numbers}")

numbers.insert(0, 2)   # Insert at index
print(f"After insert: {numbers}")

numbers.sort()         # Sort in place
print(f"Sorted: {numbers}")

numbers.reverse()      # Reverse in place
print(f"Reversed: {numbers}")
```

```python
# List comprehension - create lists concisely
squares = [x**2 for x in range(1, 6)]
print(f"Squares: {squares}")

evens = [x for x in range(10) if x % 2 == 0]
print(f"Evens: {evens}")
```

## 6. Dictionaries

Dictionaries store key-value pairs.

```python
# Create a dictionary
person = {
    "name": "Alice",
    "age": 30,
    "city": "New York"
}

print(f"Person: {person}")
print(f"Name: {person['name']}")
print(f"Age: {person.get('age')}")
```

```python
# Modify dictionary
person["email"] = "alice@example.com"  # Add new key
person["age"] = 31                      # Update existing

print(f"Updated: {person}")

# Iterate over dictionary
print("\nAll info:")
for key, value in person.items():
    print(f"  {key}: {value}")
```
